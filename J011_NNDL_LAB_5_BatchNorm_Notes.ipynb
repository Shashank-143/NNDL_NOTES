**Internal Covariance Shift** 

Covariant shift, also known as covariate shift, occurs when the distribution of the input data changes between the training and testing phases of a machine learning model, while the conditional distribution of the output given the input remains unchanged. This phenomenon can lead to a model's poor performance on new, unseen data despite performing well on training data.

Deep nets are unstable because every layer destabalizes the next

Causes:
1. Change in data distribution

Affects:
1. Slower Convergence
2. Training Instability
3. Vanishing and Exploding Gradient
4. Poor Generalization

Solution: Batch Normalization

**Batch Normalization**

The method normalizes layer inputs to be zero-mean and unit standard deviation for each training mini-batch, addressing the problem of internal covariate shift

Batch Normalization isn't a layer but it puts the normalization on individual neurons. 
It always go with mini-batch gradient. --- General RULE

Batch Norm sits between the linear step and the activation.

It allows for the use of much higher learning rates, reduces sensitivity to parameter initialization, accelerates convergence, and can eliminate the need for Dropout.

Batch Norm is not a sure-shot solution that it can be generalized everywhere. In classification problems, it would give the values in between the range that means the model is less confident in its predictions.
