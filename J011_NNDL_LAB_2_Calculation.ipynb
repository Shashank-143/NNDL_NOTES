# 2 layers

```python
x = np.array([1, 2])
W1 = np.array([[1.0, 1.5], [0.5, -1.0]])
W2 = np.array([[1.0, 1.5]])
```

```python
h = W1 @ x
h
```

```python
out_two_layers = W2 @ h
out_two_layers
```

```python
W_combined = W2 @ W1
W_combined
```

```python
out_one_layer = W_combined @ x
out_one_layer
```

```python
h_best = np.maximum(0, W1 @ x)
h_best
```

```python
out_best = W2 @ h_best
out_best
```
