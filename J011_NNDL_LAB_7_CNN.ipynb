# CNN

**Drawback of ANN:** 
- It still classifies images of cat as cat even if the pixels are rearranged and the image no longer looks like a cat anymore.
- Flattening destroys the only structure an image has. 
- A real photo bankrupts the first layer. A real pic (1000 x 1000 pixels) takes a lot of computation in the training. It can take approx 3 billion weights itself in the first layer.
- An ANN would have to learn 'cat' seperately in every corner. It doesn't generallizes well on a normal image since we are flattening it.

CNN solves this by parameter sharing by using a common filter across the nodes.
A filter is one neuron that looks at a patch and slides.

**Problems with CNN:**
- Every convolutional layer shrinks the image.
- More weightage is gien to middle pixels than corner pixels.

**Dense Layer:**

$$z^{[l]} = w^{[l]} \cdot a^{[l-1]} + b^{[l]}$$

**Convolutional Layer:**

$$z^{[l]} = w^{[l]} \ast a^{[l-1]} + b^{[l]}$$

**Activation Function:**
$$a^{[l]} = g(z^{[l]})$$

### 3. ZFNet

The most significant legacy of ZFNet is its introduction of a multi-layered Deconvolutional Network (DeconvNet). This visualization technique allowed researchers to project intermediate feature activations back into the pixel space, revealing exactly what different layers of the neural network were learning (e.g., edges in early layers, complex object parts in deeper layers).

Prior to ZFNet, CNNs were largely designed through trial and error. The creators used their DeconvNet to identify architectural bottlenecks in AlexNet, such as dead features and aliasing artifacts. They resolved these by:

- Reducing the first-layer filter size from 11×11 to 7×7 to capture more mid-frequency information.
- Reducing the first-layer convolution stride from 4 to 2 to retain more spatial pixel information.
